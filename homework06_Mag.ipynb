#### Задача 2.

С помощью модуля math создайте переменную infinity, равную математической бесконечности. Затем создайте цикл со счетчиком, который будет продолжаться до тех пор, пока значение счетчика будет меньше infinity (то есть бесконечно). Внутри цикла поместите input() и задайте условие, что если пользователь ввел 'exit', программа должна выходить из цикла по break().

```python
infinity = math.inf
count = 0
while count < infinity:
    user = input('Если хотите выйти, введите "exit"')
    count += 1
    if user.lower() == 'exit':
        break
```

#### Задача 3.

Протестируйте еще один способ создания бесконечного цикла. С помощью функции count() модуля itertools создайте бесконечную последовательности целых чисел. Печатайте элементы этой последовательности один за другим, в столбик, и после каждого запрашивайте ввод у пользователя. Как и раньше, пусть программа выходит из цикла по  break(), если пользователь ввел 'exit'.

```python
from itertools import count

for num in count(1):
    print(num)
    user = input('Если хотите выйти, введите "exit"')
    if user.lower() == 'exit':
        break

```

#### Задача 4.

Дополните код из предыдущего задания: пусть программа выходит после слова 'exit' не по break, a по команде exit(0) из модуля sys.

```python
import sys

for num in count(1):
    print(num)
    user = input('Если хотите выйти, введите "exit"')
    if user.lower() == 'exit':
        sys.exit(0)
```
