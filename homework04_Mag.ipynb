#### Задача 2.

Напишите функцию, которая будет вычислять площадь прямоугольника по его сторонам.

```python
import math
def square(a, b, c):
    p = (a + b + c) / 2
    s = math.sqrt(p * (p - a) * (p - b) * (p - c))
    return s

a = 3
b = 4
c = 5

print(square(a, b, c))
```

#### Задача 3.

Дается большое целое число, представленное как список из его цифр (например, [5, 4, 3, 2, 1] = 54321). Функция должна увеличить это число на один и вернуть его таким же списком.

```python
def change(smth):
    num = 0
    for i in range(len(smth)):
        num += smth[i] * 10 ** (len(smth) - i - 1)
    num += 1
    return [int(ch) for ch in str(num)]


data = [5, 4, 3, 2, 9]
print(change(data))
```

#### Задача 4.

```python
def figure_out(string):
    lat = 'abcdefghijklmnopqrstuvwxyz'
    for word in string.split():
        for l in word.strip('.').strip(','):
            if l not in lat:
                break
        else:
            print(word)



text = input()
figure_out(text)
```

#### Задача 5.

Панграмма - это такое предложение, в котором все буквы алфавита встречаются по одному разу.

Напишите функцию, которая будет принимать строку с предложением и проверять, что оно - панграмма.

Примеры для проверки можете взять на Википедии.

```python
def pangram(string):
	string = string.lower()
	alpha = 'абвгдеёжзийклмнопрстуфхцчшщъыьэюя'
	for sym in string:
		if sym not in alpha or string.count(sym) != 1:
			return False
	return True

PUNCT = set('!"#$%&\'()*+,./:;<=>?@[\\]^`{|}~«»—…-—')
sent = input().replace(' ', '')
for char in PUNCT:
	sent = sent.replace(char, '')
pangram(sent)
# проверила: Любя, съешь щипцы, — вздохнёт мэр, — кайф жгуч.
```

#### Задача 7 (*) - немного настоящей лингвистики.

Разберитесь, как работает функция склонения русских прилагательных из кода ниже. Расширьте ее как-нибудь, например, добавьте поддержку прилагательных с основой на -ц типа "куцый". Добавьте соответствующие кейсы в тест, чтобы проверить, что все работает правильно.

```python
def decline_adjective(masculine_form, gender, number="singular", type="qualitative"):
    """
    Правильное склонение русских прилагательных.
    """
    if masculine_form.endswith("ый") or masculine_form.endswith("ой"):
        # Твёрдое склонение
        if masculine_form.endswith("ой"):
            stem = masculine_form[:-2]
        else:  # "ый"
            stem = masculine_form[:-2]

        if number == "singular":
            if gender == "m":
                return masculine_form
            elif gender == "f":
                return stem + "ая"
            elif gender == "n":
                if stem.endswith('ц'):
                    return stem + "ее"
                else:
                    return stem + "ое"
        else:  # plural
            # После шипящих пишем "ие", иначе "ые"
            if stem.endswith(("ж", "ш", "ч", "щ")):
                return stem + "ие"
            else:
                return stem + "ые"

    elif masculine_form.endswith("ий"):
        stem = masculine_form[:-2]
        last_char = stem[-1] if stem else ""

        # лисий, волчий
        if type == "possessive":
            if number == "singular":
                if gender == "m":
                    return masculine_form
                elif gender == "f":
                    return stem + "ья"
                elif gender == "n":
                    return stem + "ье"
            else:  # plural
                return stem + "ьи"
        

        # Определяем тип основы
        elif last_char in "жшчщ":
            # Шипящие: хороший
            if number == "singular":
                if gender == "m":
                    return masculine_form
                elif gender == "f":
                    return stem + "ая"
                elif gender == "n":
                    return stem + "ее"
            else:  # plural
                return stem + "ие"

        elif last_char in "хгк":
            # Заднеязычные и х: тихий, строгий
            if number == "singular":
                if gender == "m":
                    return masculine_form
                elif gender == "f":
                    return stem + "ая"
                elif gender == "n":
                    return stem + "ое"
            else:  # plural
                return stem + "ие"


        else:
            # Мягкое склонение: синий
            if number == "singular":
                if gender == "m":
                    return masculine_form
                elif gender == "f":
                    return stem + "яя"
                elif gender == "n":
                    return stem + "ее"
            else:  # plural
                return stem + "ие"

    return masculine_form


# Тестирование
test_cases = [
    # Твёрдое склонение
    ("новый", "f", "singular", "новая"),
    ("новый", "n", "singular", "новое"),
    ("новый", "f", "plural", "новые"),
    ("большой", "f", "singular", "большая"),
    ("большой", "n", "singular", "большое"),
    ("большой", "f", "plural", "большие"),

    # Мягкое склонение (синий тип)
    ("синий", "f", "singular", "синяя"),
    ("синий", "n", "singular", "синее"),
    ("синий", "f", "plural", "синие"),

    # Шипящие (хороший тип)
    ("хороший", "f", "singular", "хорошая"),
    ("хороший", "n", "singular", "хорошее"),
    ("хороший", "f", "plural", "хорошие"),

    # Заднеязычные и х (тихий тип)
    ("тихий", "f", "singular", "тихая"),
    ("тихий", "n", "singular", "тихое"),
    ("тихий", "f", "plural", "тихие"),
    ("строгий", "f", "singular", "строгая"),
    ("строгий", "n", "singular", "строгое"),
    ("строгий", "f", "plural", "строгие"),

    ("волчий", "f", "singular", "волчья", "possessive"),
    ("волчий", "n", "singular", "волчье", "possessive"),
    ("волчий", "f", "plural", "волчьи", "possessive"),
]

for case in test_cases:
    adj, gender, number, expected, *rest = case
    if rest:
        typ = rest[0]
        result = decline_adjective(adj, gender, number, typ)
    else:
        result = decline_adjective(adj, gender, number, rest)
    status = "✓" if result == expected else f"✗ (получено {result})"
    print(f"{adj} → {result} (должно быть {expected}) {status}")
```
