```python
class Language:

    def __init__(self, name, family, living, wr_system):
        self.name = name
        self.family = family
        self.living = living
        self.wr_system = wr_system

    def __str__(self):
        return f'{self.name}: семья - {self.family}, живой - {self.living}, письменность - {self.wr_system}'


class IndoEuropean(Language):

    def __init__(self, name, family, living, wr_system):
        if living:
            wr_system = 'alphabet'
        super().__init__(name, family, living, wr_system)

```

Создайте класс предложений с атрибутами содержания и вида, в который по умолчанию пусть записывается "повествовательное". Запросите у пользователя ввод с клавиатуры любого текста, разделите этот текст на фрагменты по символам точек, восклицательных и вопросительных знаков. Затем создайте список, в который для каждого фрагмента записывайте объект класса предложений с атрибутом содержания, равным самому этому фрагменту, и атрибутом вида "вопросительное", если фрагмент оканчивается на вопросительный знак, и "побудительное/восклицательное", если фрагмент оканчивается на восклицательный знак. Затем выведите атрибуты полученных объектов.

**Пример:**

*Ввод:* Подтверждается, шо слона он купил в нашем магазине. Покупайте наших слоноу!

*Вывод:*

Подтверждается, шо слона он купил в нашем магазине - повествовательное

Покупайте наших слоноу! - побудительное/восклицательное

```python
class Sentence:

    genre = 'повествовательное'
    def __init__(self, text):
        self.text = text
        if text[-1] == '?':
            self.genre = 'вопросительое'
        elif text[-1] == '!':
            self.genre = 'побудительное/восклицательое'

    def __str__(self):
        return f'{self.text} - {self.genre}'
```

```python
punct = '.?!'
text = input('Введите текст: ')
sentences = list()
sentence = ''
for sym in text:
    sentence += sym
    if sym in punct:
        sentences.append(sentence.lstrip(' '))
        sentence = ''
sentences = [Sentence(sent) for sent in sentences]
for sent in sentences:
    print(sent)
```

```python
# Вам нужно написать функцию, которая получает на вход словарь, 
# превращает его в объект класса книг с атрибутами, соответствующими ключам словаря, 
# и записывает в эти атрибуты значения по соответствующим ключам. 
# Пройдите этой функцией по всему данному вам списку словарей, записывая объекты, 
# получившиеся в результате, также в список. 
# То есть в конечном счете у вас должен получиться список объектов, 
# равный по длине изначальному списку словарей.

class Book:

    def __init__(self, title, author, year, genre, tokens, unique_words, top_words):
        self.title = title
        self.author = author
        self.year = year
        self.genre = genre
        self.tokens = tokens
        self.unique_words = unique_words
        self.top_words = top_words
    # id, title, author(str), year(int), genre(str), tokens(int), unique_words(int), top_words(dict)
    def __str__(self):
        data = f'{self.title}\nАвтор: {self.author}\nГод: {self.year}\nЖанр: {self.genre}\nВсего уникальных слов: {self.unique_words}\nЧастотные слова:\n'
        for w, num in self.top_words.items():
            data += f'{w:<3}: {num}\n'
        return data


def creating_book(dicti):
    id, *remaining = dicti.values()
    book = Book(*remaining)
    return book

def creating_books(whole):
    books = [creating_book(book) for book in whole]
    return books

corpus_update = creating_books(corpus)

for book in corpus_update:
    print(book)
```
