#### Задача 7 (*) - байты и сигнатуры.

Ниже дан код функции-файлчекера, которая проверяет байтовую сигнатуру файла на совпадение с известными форматами. Возьмите какой-нибудь нетекстовый файл и переименуйте его вручную в формат .txt, а затем дополните код ниже так, чтобы программа попыталась открыть его через обычный open. Посмотрите, что будет. После этого закомментируйте строку с open и примените к этому файлу функцию check_file_type. Выведите на экран результат. Напишите в комментариях подробно, как вы понимаете работу этой функции.

```python
def check_file_type(file_path):

    signatures = {
        b'\x89PNG\r\n\x1a\n': 'PNG Image',
        b'\xff\xd8\xff': 'JPEG Image',
        b'%PDF': 'PDF Document',
        b'PK\x03\x04': 'ZIP archive (or Office Document)'
    }

    with open(file_path, 'rb') as f:
        header = f.read(8)

    for sig, file_type in signatures.items():
        if header.startswith(sig):
            return f"Это реальный {file_type}"

    return "Неизвестный тип файла"

# your code here
```
