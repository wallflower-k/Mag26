#### Задача 1.

Нужно открыть любой текстовый файл и считать из него текст в виде списка слов без пунктуации. Далее нужно записать получившийся список слов по слову на строчке в новый файл.

```python
def punct(item):
    punctuation = '!.,?:;-_—…»«'
    for sym in punctuation:
        item = item.strip(sym)
    return item

with open('приколюха.txt', 'r', encoding='utf8') as file:
    text = file.read()


cleared_text = [punct(i) for i in text.split() if punct(i)]
```

```python
print(cleared_text)
```

```python
with open('приколюха_update.txt', 'w', encoding='utf8') as text_update:
    text_update.write('\n'.join(cleared_text))
```

#### Задача 2.

Дополните код из предыдущего задания. Прочитайте созданный вами файл построчно в цикле. Вам нужно узнать среднюю длину только тех слов из файла, которые длинее трех букв, и вывести результат на экран.

**Примечание:** Помните про символы \n, которые не являются частью слов!

```python
count = 0
summ = 0
with open('приколюха_update.txt', 'r', encoding='utf8') as new_text:
    for string in new_text:
        if len(string) > 3:
            summ += len(string.strip('\n'))
            count += 1

total = summ / count
total
```

```python
def check_file_type(file_path):

    signatures = {
        b'\x89PNG\r\n\x1a\n': 'PNG Image',
        b'\xff\xd8\xff': 'JPEG Image',
        b'%PDF': 'PDF Document',
        b'PK\x03\x04': 'ZIP archive (or Office Document)'
    }

    with open(file_path, 'rb') as f:
        header = f.read(8)

    for sig, file_type in signatures.items():
        if header.startswith(sig):
            return f"Это реальный {file_type}"

    return "Неизвестный тип файла"


with open('sem07/wallpaper.txt', 'r') as filejpg:
    textik = filejpg.read()
print(textik)
```
