```python
def cyrillic(word):
  cyr = {'а', 'б', 'в', 'г', 'д', 'е', 'ё', 'ж', 'з', 'и', 'й', 'к', 'л', 'м', 'н', 'о', 'п', 'р', 'с', 'т', 'у', 'ф', 'х', 'ц', 'ч', 'ш', 'щ', 'ъ', 'ы', 'ь', 'э', 'ю', 'я'}
  for sym in word:
    if sym not in cyr:
      return False
  return True

print(cyrillic('машина'))
print(cyrillic('mother'))
```

#### Задача 3.

Напишите функцию для подсчета букв в слове: пользователь вводит какую-то строку, а функция должна в форматированном виде вывести, сколько раз встретилась та или иная буква.

**Пример:**

*Ввод:* 'ABBYY'

*Вывод:*

'A': 1

'B': 2

'Y': 2

```python
def count(word):
  data = dict()
  for sym in word:
    if sym in data.keys():
      data[sym] += 1
    else:
      data[sym] = 1
  return data

res = count(input())
for k, v in res.items():
  print(f'\'{k}\': {v}')
```

#### Задача 8 (*) - свойства чисел.

У некоторых целых чисел есть забавные свойства, например:

89 --> 8¹ + 9² = 89 * 1

695 --> 6² + 9³ + 5⁴= 1390 = 695 * 2

46288 --> 4³ + 6⁴+ 2⁵ + 8⁶ + 8⁷ = 2360688 = 46288 * 51

Нужно написать функцию, которая будет проверять, есть ли у натурального числа n такие свойства, а именно если цифры этого числа возвести в степени p, p + 1, p + 2..., то в результате будет число, которое делится на исходное число нацело.

Напишите функцию dig_pow(n, p), где n - само число на проверку, а p - натуральное число, с которого мы начинаем возводить цифры нашего числа в степень. Функция должна вернуть False, если наше число n таким свойством не обладает, и True, если да.

**Примеры:**

dig_pow(89, 1) должно вернуть True, потому что 8¹ + 9² = 89 = 89 * 1
dig_pow(92, 1) должно вернуть False потому что нет такого k, чтобы 9¹ + 2² равнялось 92 * k
dig_pow(695, 2) должно вернуть True, потому что 6² + 9³ + 5⁴= 1390 = 695 * 2
dig_pow(46288, 3) должно вернуть True, потому что 4³ + 6⁴+ 2⁵ + 8⁶ + 8⁷ = 2360688 = 46288 * 51



```python
def dig_pow(num, z):
    res = 0
    for i in str(num):
        res += int(i) ** z
        z += 1
    if res // num:
        return True
    return False

num, check = int(input()), int(input())
dig_pow(num, check)
```

#### Задача 9 (**) - счастливые числа

В теории чисел счастливое число (happy number) — это целое число, которое, когда его заменяют суммой квадратов каждой цифры, а получившееся затем число заменяют суммой квадратов его цифр, и т.д., в конечном итоге достигает 1. Например, 13 — счастливое число, потому что 1² + 3² = 10 и 1² + 0² = 1. С другой стороны, 4 не является счастливым числом, потому что последовательность, начинающаяся с 4, проходит через 16, 37 и т.д. и в конечном итоге снова достигает 4.

Напишите функцию, которая принимает на вход число и печатает все счастливые числа, меньшие или равные своего аргумента.

```python
def checking(num):
    summ = 0
    for ch in str(num):
        summ += int(ch) ** 2
    return summ

def happy_number(num):
    happ = list()
    for n in range(num, 1, -1):
        summ = checking(n)
        while summ > 6: #так как 7 наименьшее счастливое число
            summ = checking(summ)
        if summ == 1:
            happ.append(n)
    return happ

happy_number(32)
```
