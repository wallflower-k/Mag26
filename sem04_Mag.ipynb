#### Задача 2.

Напишите функцию, которая будет принимать слово и проверять, состоит ли оно только из кириллических букв. Функция должна возвращать объект bool (True/False).

```python
def cyrillic(word):
  cyr = {'а', 'б', 'в', 'г', 'д', 'е', 'ё', 'ж', 'з', 'и', 'й', 'к', 'л', 'м', 'н', 'о', 'п', 'р', 'с', 'т', 'у', 'ф', 'х', 'ц', 'ч', 'ш', 'щ', 'ъ', 'ы', 'ь', 'э', 'ю', 'я'}
  for sym in word:
    if sym not in cyr:
      return False
  return True

print(cyrillic('машина'))
print(cyrillic('mother'))
```

#### Задача 3.

Напишите функцию для подсчета букв в слове: пользователь вводит какую-то строку, а функция должна в форматированном виде вывести, сколько раз встретилась та или иная буква.

**Пример:**

*Ввод:* 'ABBYY'

*Вывод:*

'A': 1

'B': 2

'Y': 2

```python
def count(word):
  data = dict()
  for sym in word:
    if sym in data.keys():
      data[sym] += 1
    else:
      data[sym] = 1
  return data

res = count(input())
for k, v in res.items():
  print(f'\'{k}\': {v}')
```
