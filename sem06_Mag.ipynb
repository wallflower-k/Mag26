#### Задача 1.

Возьмите код какой-нибудь из прошлых задач, которую вы решали. С помощью модуля time измерьте время работы этого кода и выведите его на экран.

```python
import time

start = time.time()
call = ['звать', 'звонить', 'называть']
name = ['имя', 'называть']
eng_rus = dict()
eng_rus['call'] = call
eng_rus['name'] = name
print(eng_rus)

rus_eng = dict()
for word in eng_rus:
    for trans in eng_rus[word]:
        if trans not in rus_eng:
            rus_eng[trans] = word
        else:
            temp = list()
            temp.append(rus_eng[trans])
            temp.append(word)
            rus_eng[trans] = sorted(temp)
for k in sorted(rus_eng.keys()):
    print(f'{k}: {rus_eng[k]}')
end = time.time()

print(end - start)
```

#### Задача 4.

С помощью модуля datetime создайте таймер, который можно установить на заданное время, и он будет показывать обратный отсчет (с точностью до секунд).

```python
from datetime import datetime
user_timer = int(input('На сколько завести таймер? (в секундах)'))
target = datetime.now() + timedelta(seconds=user_timer)
while target > datetime.now():
    print(target - datetime.now())
    time.sleep(1)
print('ВСЁ')
```

#### Задача 5.

С помощью модуля random выберите случайное число от 1 до 10 (не выводите его, а просто запишите в переменную). Затем предложите пользователю угадать с трех раз, что это за число. Если он не угадает с первой попытки, напишите "больше" или "меньше" в зависимости от того, какое число загадано. Если он не угадает со второго раза, снова напишите ему подсказку и как-нибудь подбодрите его. В случае выигрыша поздавьте его, а в случае проигрыша предложите сыграть еще раз.

```python
from random import randint

def guessing(num):
    for _ in range(3):
        user_guess = int(input('Угадайте число от 1 до 10!'))
        if user_guess == num:
            return 'Поздравляю вы угадали!'
        elif user_guess > num:
            print('Меньше')
        else:
            print('Больше')
    return 'К сожалению вы не угадали. Хотите сыграть еще раз?'

something = randint(1,10)
print(guessing(something))
```

#### Задача 7 (*) - комбинаторика.

Разберитесь, как работает модуль itertools. С его помощью решите следующую задачу: нужно подсчитать число способов, которыми *n* писем можно разложить по *m* ящикам. Числа *n* и *m* вводятся пользователем. Гарантируется, что *n* не превышает *m*.

```python
from itertools import product

n = int(input('Введите количество писем: '))
m = int(input('Введите количество ящиков: '))

count = 0
for _ in product(range(m), repeat=n):
    count += 1
print(count)
```
