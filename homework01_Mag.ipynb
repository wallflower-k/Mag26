#### Задача 1.

Дано трехзначное число. Нужно узнать, есть ли в нем хотя бы одна цифра 7. Если такая цифра есть, выведите на экран "Удача!", в ином случае "Неудача". Работайте с числом как с int, не в виде строки.


**Подсказка:** Алгоритм выделения цифр трехзначного числа можно списать из текста лекции.

```python
num = int(input())

if num % 10 == 7 or num % 100 // 10 == 7 or num // 100 == 7:
    print('Удача!')
else:
    print('Неудача')
```

#### Задача 3.

Напишите программу, которая принимает на вход число секунд и возвращает ответ в часах, минутах и секундах в виде 12:13:14.

**Подсказка:** можно использовать остаток от деления и целочисленное деление, а также параметр end функции print().

```python
time = int(input('Введите количество секунд: '))

sec = time % 60
minut = time // 60 % 60
hours = time // 60 // 60

print(f'{hours}:{minut}:{sec}')
```

#### Задача 4.

Напишите программу, которая переводит вес из килограммов в фунты. Вес вводится пользователем и может быть не целым.

**Примечание:** Фунт делится на 16 унций, а не на 10, так что если будете выводить дробную часть фунта, выводите в правильных единицах!

```python
weight = float(input())

lb = weight * 2.20462

dict_lb = {0: 'фунтов', 1: 'фунт', 2: 'фунта', 3: 'фунта', 4: 'фунта', 5: 'фунтов', 6: 'фунтов', 7: 'фунтов', 8: 'фунтов', 9: 'фунтов'}
dict_oz = {0: 'унций', 1: 'унция', 2: 'унции', 3: 'унции', 4: 'унции', 5: 'унций', 6: 'унций', 7: 'унций', 8: 'унций', 9: 'унций'}

oz = lb % 1 * 16

print(f'{int(lb)} {dict_lb[lb // 1 % 10]} {round(oz, 2)} {dict_oz[oz % 10 // 1]}')
```

#### Задача 5.

Точки А, В, С расположены на одной прямой в том порядке, в котором они написаны (т.е. точка В расположена между точками А и С). Из точки В в точку С со скоростью v едет велосипедист, а из точки А со скоростью u его догоняет мотоциклист. Напишите программу, которая запрашивает у пользователя расстояния АВ и ВС, скорости v и u, а в качестве ответа выдает "Да", если мотоциклист догонит велосипедиста до того, пока тот не достиг точки С, и "Нет" в противном случае.

```python
ab = int(input('расстояние АВ (в метрах): '))
bc = int(input('рассотяние ВС (в метрах): '))

v = int(input('скорость велосипедиста (в км/ч): '))
u = int(input('скорсоть мотоциклиста (в км/ч): '))

ab //= 1000
bc //= 1000

time_v = bc / v
time_u = (ab + bc) / u

if time_u < time_v:
    print('Да')
else:
    print('Нет')
```

#### Задача 6.

Напишите простой калькулятор. Пользователь вводит два числа (целых или дробных) и символ операции (+, -, *, /). Программа должна вывести результат вычисления. Если пользователь ввел /, а второе число равно 0, программа должна вывести «Ошибка: деление на ноль!».

**Примечание:** Символы операций обрабатываются питоном точно так же, как буквы.

```python
num1 = float(input())
num2 = float(input())

oper = input('введите +, -, *, /')

if num2 == 0 and oper == '/':
    print('Ошибка: деление на ноль!')
elif oper == '+':
    print(num1 + num2)
elif oper == '-':
    print (num1 - num2)
elif oper == '*':
    print(num1 * num2)
elif oper == '/':
    print(num1 / num2)
else:
    print('Ошибка: неправильный ввод')
```
