#### Задача 1.

Собираем телефонный справочник. Пользователь в бесконечном цикле вводит фамилию и номер телефона через пробел, пока не введет "хватит". Программа должна завести словарь, в котором по фамилии хранится номер телефона, и вывести его на экран.

```python
tel_book = dict()
while True:
    surname = input('surname: ')
    if surname == 'хватит':
        break
    tel_book[surname] = input('number: ')
    print(tel_book)
```

У нас есть два списка: в одном по порядку записаны имена людей, а в другом их дни рождения (формат неважен, можно в строках). Нам нужно объединить два списка в словарь, где ключ - имя, а значение - день рождения, причем мы знаем, что дни рождения в том же порядке, что и имена людей. После этого пусть программа запросит у пользователя имя человека и выдаст ему день рождения этого человека, либо сообщение "Запись отсутствует", если такого человека нет в словаре.

```python
list1 = ['Ксюша', 'Саша', 'Валя']
list2 = ['21.07', '06.10', '01.06']
base_dict = dict()
for l1, l2 in zip(list1, list2):
    base_dict[l1] = l2

user_request = input('Введите имя: ')
print(base_dict.get(user_request, 'Запись отсутствует'))
```

#### Задача 3.

Вводится текст. Нам нужно посчитать относительные частоты всех его слов: то есть, частоты слов нужно дополнительно поделить на общее их количество. В результате программа должна выдать словарь, где слово является ключом, а его частота - значением. Для выдачи используйте метод .items()

```python
text = 'Ввиду этой богомногочисленности самому осторожному и набожному египтянину ежеминутн приходилось совершать различные кощунства. То наступит кошке на хвост, то цыкнет на священную собаку, то съест в борще святую муху.'

data = [word.lower().strip(',').strip('.') for word in text.split()]
res = {word: (data.count(word) / len(data)) for word in set(data)}
for word, count in res.items():
    print(f'{word}: {count}')
```

#### Задача 4.

Даны два словаря, и известно, что в них могут встречаться записи со значением None. Нужно объединить их, но так, чтобы в результирующий словарь из двух записей с одинаковыми ключами и разными значениями попала та, значение которой отлично от None. #### Задача 5

Превратим англо-русский словарь в русско-английский. В исходном англо-русском словаре для одного английского слова могут идти в алфавитном порядке русские варианты перевода:

    'call': ['звать', 'звонить', 'называть']

Может быть и такая ситуация, что одно и то же русское слово встречается в переводах двух английских слов:

    'name': ['имя', 'называть']

Наш русско-английский словарь должен правильно перевернуться: не потерять ни одного перевода, при этом английские переводы тоже должны быть в алфавитном порядке. Программа должна вывести новый словарь на экран, причем русские слова должны быть отсортированы, как и их переводы.

```python
call = ['звать', 'звонить', 'называть']
name = ['имя', 'называть']
eng_rus = dict()
eng_rus['call'] = call
eng_rus['name'] = name
print(eng_rus)

rus_eng = dict()
for word in eng_rus:
    for trans in eng_rus[word]:
        if trans not in rus_eng:
            rus_eng[trans] = word
        else:
            temp = list()
            temp.append(rus_eng[trans])
            temp.append(word)
            rus_eng[trans] = sorted(temp)
for k in sorted(rus_eng.keys()):
    print(f'{k}: {rus_eng[k]}')
```
