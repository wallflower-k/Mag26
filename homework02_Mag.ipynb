#### Задача 3.

Дана строка. Если в этой строке буква f встречается только один раз, выведите её индекс. Если она встречается два и более раз, выведите индекс её первого и последнего появления. Если буква f в данной строке не встречается, ничего не выводите.

```python
string = input('введите строку: ')
f_amt = string.count('f')
if f_amt == 1:
    print(string.find('f'))
elif f_amt > 1:
    print(string.find('f'), string.rfind('f'))
```

#### Задача 7.

Пользователь в цикле вводит сочетание двух слов. Если оба слова начинаются на согласные, программа должна поменять местами их части до первой гласной и вывести получившееся словосочетание. Если одно или оба слова начинаются на гласные, то программа должна поменять местами части слов до буквы, следующей за первой гласной. Концом ввода считается словосочетание, в котором есть хотя бы одно слово, не содержащее гласных букв.

**Примеры:** плоды трудов -> троды плудов; Омар Хаям -> Хамар Оям

```python
def is_vowel(sym):
    vowels = 'аеёиоуыэюяАЕЁИОУЫЭЮЯ'
    if sym in vowels:
        return True
    return False


def whole_word(word):
    for sym in word:
        if is_vowel(sym):
            return True
    return False

def first_vowel(word):
    for ind, sym in enumerate(word):
        if is_vowel(sym):
            return ind


two_wordsss = input('Введите словосочетание: ')
two_words = two_wordsss.split()
while whole_word(two_words[0]) and whole_word(two_words[1]):
    if (not is_vowel(two_words[0][0])) and (not is_vowel(two_words[1][0])):
        print(two_wordsss, '->', end=' ')
        v1, v2 = first_vowel(two_words[0]), first_vowel(two_words[1])
        w1, w2 = two_words[1][:v2] + two_words[0][v1:], two_words[0][:v1] + two_words[1][v2:]
        print(w1, w2)
    else:
        print(two_wordsss, '->', end=' ')
        v1, v2 = first_vowel(two_words[0]) + 1, first_vowel(two_words[1]) + 1
        w1, w2 = two_words[1][:v2] + two_words[0][v1:], two_words[0][:v1] + two_words[1][v2:]
        print(w1, w2)
    two_wordsss = input('Введите словосочетание: ')
    two_words = two_wordsss.split()
```

#### Задача 8 (*) - шифр Цезаря

Напишите программу, которая сдвигает буквы в строке на *n* позиций вперед по алфавиту. Сдвиг циклический, т.е. если если достигнута последняя буква алфавита, то следующей будет первая буква, потом вторая и т.д. Пробелы и знаки препинания символами алфавита не считаются. Строка и число *n* вводятся пользователем. Регистр букв не имеет значения.

```python
def shifting(string, n):
    # dict_cyr_cap = {sym: ord(sym) for sym in 'АБВГДЕЁЖЗИЙКЛМНОПРСТУФХЦЧШЩЪЫЬЭЮЯ' if ord(sym) <= ord('Е') else sym: ord(sym) + 1}
    dict_cyr = {sym: ord(sym) if ord(sym) <= ord('е') else ord(sym) + 1 for sym in 'абвгдеёжзийклмнопрстуфхцчшщъыьэюя'}
    dict_cyr['ё'] = ord('е') + 1
    res = ''
    for letter in string:
        if letter.isalpha():
            if letter.isupper():
                letter = letter.lower()
                flag = True
            else:
                flag = False
            if letter.islower():
                num = dict_cyr[letter] + n
                if num > dict_cyr['я']:
                    num = num - dict_cyr['я'] + dict_cyr['а']
            sym = [k for k, v in dict_cyr.items() if v == num][0]
            if flag:
                sym = sym.upper()
            res += sym
        else:
            res += letter
    return res


shift = int(input('Введите число, относительно которого будут сдвинуты буквы: '))

sentence = 'Ввиду этой богомногочисленности самому осторожному и набожному египтянину ежеминутн приходилось совершать различные кощунства. То наступит кошке на хвост, то цыкнет на священную собаку, то съест в борще святую муху.'

print(shifting(sentence, shift), '\nСдвиг:', shift)
```
