#### Задача 5.

Дополните код из предыдущего задания: создайте вторую сиамскую кошку и напишите функцию fight, которая будет принимать на вход два объекта класса кошек и проверять принадлежность их к сиамским. Если обе кошки сиамские, то пусть функция выводит через запятую их клички, а затем сообщение "Хорошая драка!" Если только одна из кошек сиамская, то пусть функция выводит кличку второй (не сиамской) кошки, а затем сообщение "покидает поле боя", и удаляет объект. Подумайте, как надо дополнить класс кошек, чтобы сообщение об удалении проигравшей в драке кошки вызывалось не самой функцией, а методом этого класса. Если обе кошки не сиамские, пусть функция выводит сообщение "Драки нет, мир да любовь.", и больше ничего не делает.

```python
def fight(c1, c2):
    if isinstance(c1, Siam) == isinstance(c2, Siam):
        if isinstance(c1, Siam):
            print(f'{c1.name}, {c2.name}\nХорошая драка!')
        else:
            print('Драки нет, мир да любовь.')
    else:
        if isinstance(c1, Siam):
            c2.delete()
        else:
            c1.delete()


class Cat:

    def __init__(self, name):
        self.name = name

    def delete(self):
        print(f'{self.name} покидает поле боя.')

    def purr(self):
        print('Мррр')


class Siam(Cat):

    def __init__(self, name, eyes='голубой'):
        super().__init__(name)
        self.eyes = eyes


kitten = Siam('Чоп', 'зелёный')

second_kitten = Cat('Пушистик')

fight(kitten, second_kitten)
```

```python
def fight(c1, c2):
    if isinstance(c1, Siam) == isinstance(c2, Siam):
        if isinstance(c1, Siam):
            return f'{c1.name}, {c2.name}\nХорошая драка!'
        else:
            return f'Драки нет, мир да любовь.'
    else:
        if isinstance(c1, Siam):
            return f'{c2.name} покидает поле боя.'
            c2.delete()
        else:
            return f'{c1.name} покидает поле боя.'
            c1.delete()


class Cat:

    def __init__(self, name):
        self.name = name

    def delete(self):
        print(f'{self.name} покидает поле боя.')

    def purr(self):
        print('Мррр')


class Siam(Cat):

    def __init__(self, name, eyes='голубой'):
        super().__init__(name)
        self.eyes = eyes


kitten = Siam('Чоп', 'зелёный')

second_kitten = Cat('Пушистик')

fight(kitten, second_kitten)

cat1 = Siam('Чоп')
cat2 = Siam('Рыжик')
cat3 = Cat('Пушистик')
cat4 = Cat('Мурка')
```

#### Задача 7.

Задайте теперь базовый класс собак. Пусть атрибутами собак будут кличка и размер (крупный, средний, мелкий). Методом класса собак пусть будет рост: при вызове этого метода мелкая собака пусть превращается в среднюю, а средняя в крупную. Создайте несколько объектов этого класса, выведите их атрибуты на экран, затем вызовите у них этот метод и снова выведите их атрибуты.

```python
class Dog:

    def __init__(self, name, size):
        self.name = name
        self.size = size

    def __str__(self):
        return f"Кличка: {self.name}, размер: {self.size}"

    def height(self):
        if self.size == 'мелкий':
            self.size = 'средний'
        elif self.size == 'средний':
            self.size = 'крупный'


dog1 = Dog('Лимон', 'мелкий')
dog2 = Dog('Тедди', 'средний')
dog3 = Dog('Несси', 'крупный')
```

```python
for dog in (dog1, dog2, dog3):
    print(dog)
```

```python
for dog in (dog1, dog2, dog3):
    dog.height()

for dog in (dog1, dog2, dog3):
    print(dog)
```

#### Задача 8 (*) - декорирование

Создайте декоратор для класса собак, который будет изменять метод роста так, чтобы при каждом изменении атрибута размера метод выводил сообщение "Собака выросла!", а затем кличку собаки и ее новый размер.

```python
def my_decorator(func):
    def wrapper(self):
        old_size = self.size
        func(self)
        if self.size != old_size:
            print('Собака выросла!')
            print(self)
    return wrapper
    
class Dog:

    def __init__(self, name, size):
        self.name = name
        self.size = size

    def __str__(self):
        return f"Кличка: {self.name}, размер: {self.size}"

    @my_decorator
    def height(self):
        if self.size == 'мелкий':
            self.size = 'средний'
        elif self.size == 'средний':
            self.size = 'крупный'


dog1 = Dog('Лимон', 'мелкий')
dog2 = Dog('Тедди', 'средний')
dog3 = Dog('Несси', 'крупный')
```

```python
for dog in (dog1, dog2, dog3):
    print(dog)

for dog in (dog1, dog2, dog3):
    dog.height()
```
