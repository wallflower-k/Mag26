#### Задача 2.

Пишем продвинутый справочник контактов. Создайте словарь, где ключ — имя, а значение — вложенный словарь с данными (телефон, email, адрес). Реализуйте функции добавления, удаления и поиска контактов.

```python
def adding():
    global dict_tel
    name = input('Введите имя: ')
    telephone = input('Введите номер телефона: ')
    email = input('Введите e-mail: ')
    address = input('Введите адрес: ')
    dict_tel[name] = {'telephone': telephone, 'e-mail': email, 'address': address}

def deleting():
    global dict_tel
    name = input('Введите имя контакта, которое собираетесь удалить: ')
    if dict_tel.get(name):
        del dict_tel[name]
        print(f'Контакт {name} удален!')
    else:
        print('Такого контакта и не было в справочнике')

def searching():
    global dict_tel
    name = input('Введите имя контакта, который вы ищете: ')
    if dict_tel.get(name):
        print(f'{name}: \t{dict_tel[name]['telephone']}\n\t\t{dict_tel[name]['e-mail']}\n\t\t{dict_tel[name]['address']}')
    else:
        print('Нет контакта с таким именем.')

dict_tel = {}
while True:
    print('Введите 0, если хотите выйти\n1 - добавить запись в справочник\n2 - удалить контакт\n3 - найти контакт в справочнике')
    user = int(input())
    if user == 0:
        break
    elif user == 1:
        adding()
    elif user == 2:
        deleting()
    elif user == 3:
        searching()
    print(dict_tel)
```

#### Задача 4.

Напишите программу, которая подсчитывает количество единиц товаров, приобретенных покупателями онлайн-магазина. На вход программе подается число n – количество записей о покупках, а затем n строк вида «Покупатель Товар Количество». Программа должна сделать словарь, в котором имена покупателей будут ключами, а значениями вложенные словари покупок вида Товар: Количество. Красиво выведите полученную структуру на экран.

```python
sells = int(input('Введите количество покупок'))
total = dict()
for i in range(sells):
    print('Введите в формате "Покупатель Товар Количество"')
    data = input().split()
    total[data[0]] = {data[1]: data[2]}
for buyer, res in total.items():
    print(buyer)
    for k, v in res.items():
        print(f'\t{k}:\t{v}')
```

#### Задача 6

Делаем программу, которая принимает на вход слог русского языка (записанный кириллицей с употреблением спецсимволов при необходимости - напишите инструкцию для пользователя по способу записи и выведите ее на экран). В качестве результата программа выдает словарь разбора этого слога по признакам: количество букв в нем, открытость/закрытость, твердость/мягкость. При определении твердости и мягкости для простоты будем ориентироваться только на наличие мягкого знака или специального символа мягкости - апострофа.

```python
def letters(string):
    count = 0
    for sym in string:
        if sym.isalpha():
            count += 1
    return count

def isvowel(string):
    vowels = 'аеёиоуыэюя'
    if string[-1].lower() in vowels:
        return 'open'
    return 'closed'

def palatalization(string):
    if 'ь' in string or '\'' in string:
        return True
    return False


syllable = input('Введите слог: ')
analysis = {'letters': 0, 'open/closed': '', 'palat': False}
analysis['letters'] = letters(syllable)
analysis['open/closed'] = isvowel(syllable)
analysis['palat'] = palatalization(syllable)
print(analysis)
```
